# Convertendo arquivos .csv para FASTA.

```python
import csv
import os

def converter_para_fasta(csv_file_path, fasta_file_path):
    """
    Converte um arquivo CSV (ou TSV) para FASTA, criando um ID único
    a partir do número da linha e das colunas 'Cromossomo' e 'Classe'.
    
    Assume que as colunas de sequência são:
    - 'Cromossomo'
    - 'Sequência de TE'
    - 'Classe'
    """
    
    # Verifica se o arquivo de entrada existe
    if not os.path.exists(csv_file_path):
        print(f"--- ERRO ---")
        print(f"Arquivo de entrada não encontrado em: {csv_file_path}")
        print("Verifique se o caminho está correto (começando com '../data/...')")
        return

    # Tenta detectar o separador (vírgula ou TAB)
    # Se fosse um TSV, seria sep='\t'
    sep = ',' 

    print("Iniciando conversão...")
    
    try:
        with open(csv_file_path, 'r', newline='', encoding='utf-8') as infile, \
             open(fasta_file_path, 'w') as outfile:
            
            # Ajuste para usar o separador correto
            reader = csv.DictReader(infile, delimiter=sep)
            
            # Itera com 'enumerate' para obter um índice (i) para cada linha
            for i, row in enumerate(reader):
                
                # Pega os dados das colunas que você identificou
                sequence = row.get('Sequência de TE')
                cromossomo = row.get('Cromossomo')
                classe = row.get('Classe')

                # --- Criação do ID ---
                # Vamos criar um ID único e descritivo
                # Exemplo: >ID_0_Chr1_LTR
                # 'i' garante que o ID seja sempre único
                novo_id = f"ID_{i}_{cromossomo}_{classe}"
                
                # Garante que temos uma sequência antes de escrever
                if sequence:
                    outfile.write(f">{novo_id}\n")
                    outfile.write(f"{sequence}\n")
                else:
                    print(f"Aviso: Linha {i} pulada (sequência não encontrada).")

    except FileNotFoundError:
        print(f"ERRO: Arquivo de entrada não encontrado em {csv_file_path}")
        return
    except KeyError as e:
        print(f"ERRO: A coluna {e} não foi encontrada no CSV.")
        print("Verifique se os nomes das colunas no código (ex: 'Sequência de TE') estão corretos.")
        return
    except Exception as e:
        print(f"Ocorreu um erro inesperado: {e}")
        return

    print(f"Sucesso! Arquivo FASTA salvo em: {fasta_file_path}")

# --- Como usar a função ---

# Caminho do seu CSV (assumindo que está na pasta notebooks/)
# Ajuste o nome do arquivo para o seu CSV de treino
input_csv = '/home/gabyl/projetos/trabalho-TEsClassification/data/train_test_split/train_dataset_balanced.csv'

# Caminho onde o FASTA será salvo
output_fasta = '/home/gabyl/projetos/trabalho-TEsClassification/data/train_test_split_fasta/sequences_train.fasta' # Mudei o nome para ser específico

# Chama a função
converter_para_fasta(input_csv, output_fasta)
```
